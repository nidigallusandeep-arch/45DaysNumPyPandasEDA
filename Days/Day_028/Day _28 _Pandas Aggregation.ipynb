```python
# Total revenue by city
total_revenue=df.groupby("City")["Revenue_INR"].sum()
total_revenue
```

```python
# Average revenue by city
avg=df.groupby("City")["Revenue_INR"].mean()
avg
```

```python
# Total quantity by product
total=df.groupby("Product")["Quantity"].sum()
total
```

```python
# Highest price by product
highest=df.groupby("Product")["Unit_Price_INR"].max()
highest
```

```python
# Order count by status
count=df.groupby("Order_Status")["Order_ID"].count()
count
```

```python
# Revenue by category
df.groupby("Category")["Revenue_INR"].sum()
```

```python
# Revenue and quantity by city
df.groupby("City").agg(
    Total_Revenue=("Revenue_INR", "sum"),
    Total_Quantity=("Quantity", "sum"))
```

```python
# Multiple metrics by product
df.groupby("Product").agg(
    Total_Revenue=("Revenue_INR", "sum"),
    Average_Revenue=("Revenue_INR", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Order_Count=("Order_ID", "count"),
    Highest_Price=("Unit_Price_INR", "max"))
```

```python
# Multiple metrics by order status
df.groupby("Order_Status").agg(
    Order_Count=("Order_ID", "count"),
    Total_Revenue=("Revenue_INR", "sum"),
    Average_Revenue=("Revenue_INR", "mean"))
```

```python
# Find products with highest total revenue
df.groupby("Product")["Revenue_INR"].sum().sort_values(ascending=False)
```

```python
# Create a city-wise sales report
sales_report = df.groupby("City", as_index=False).agg(
    Total_Revenue=("Revenue_INR", "sum"),
    Average_Revenue=("Revenue_INR", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Order_Count=("Order_ID", "count"))
sales_report
```
