# 2. What is a Pandas Series?
- A Series is a one-dimensional labeled array. It contains values and an index.

```python
import pandas as pd
marks=pd.Series([50,60,70,80,98])
print(marks)
```

- Left side (0, 1, 2...) → index.
- Right side (85, 90...) → values.
- dtype: int64 → data type of the values.

```python
import pandas as pd
prices=pd.Series([100,300,400,500])
print(prices)
print("First price:",prices[0])
print("Total:", prices.sum())
print("Average:", prices.mean())
```

```python
marks = pd.Series(
    [85, 90, 78],
    index=["Sandeep", "Ravi", "Asha"])

print(marks)
print("Sandeep's marks:", marks["Sandeep"])
```

# 3. What is a Pandas DataFrame?
- A DataFrame is a two-dimensional table containing rows and columns. Different columns can have different data types.

```python
import pandas as pd
data={
    "name":["Sandeep","Yashu","Ravi","Sanju","Sai"],
    "age":[23,22,20,21,22],
    "marks":[90,95,80,78,90]
}
df=pd.DataFrame(data)
```

```python
df
```

# Understand the code
- data is a Python dictionary.
- Dictionary keys become column names.
- Dictionary lists become column values.
- pd.DataFrame(data) creates the table.
- print(df) displays the DataFrame.

# 4. Access rows and columns
- Select one column

```python
print(df["name"])
```

```python
# Select multiple columns
print(df[["name","marks"]])
```

```python
# Select a row using iloc
print(df.iloc[0])
```

```python
print(df.iloc[1])
```

- iloc[0] selects the first row using its integer position.

```python
# Select a particular value
print(df.iloc[0,2])
```

- Here, 0 is the first row and 2 is the third column.

```python
print(df.iloc[0,1])
```

# 5. Useful DataFrame properties

```python
print("Shape:", df.shape)
print("Rows and columns:", df.shape)
print("Column names:", df.columns)
print("Index:", df.index)
print("Data types:")
print(df.dtypes)
```

# Remember:
- shape → number of rows and columns.
- columns → column names.
- index → row labels.
- dtypes → data type of each column.

# 7. Real-time example: Employee analysis

```python
employees = pd.DataFrame({
    "Employee": ["Ravi", "Asha", "Kiran", "Neha"],
    "Department": ["IT", "HR", "IT", "Sales"],
    "Salary": [40000, 35000, 50000, 30000})

print(employees)
print("Average salary:", employees["Salary"].mean())

print("Highest salary:", employees["Salary"].max())

print("Total salary:", employees["Salary"].sum())
```

```python
employees
```

```python
# Find employees earning more than ₹35,000:
salary=employees[employees["Salary"]>35000]
print(salary)
```
