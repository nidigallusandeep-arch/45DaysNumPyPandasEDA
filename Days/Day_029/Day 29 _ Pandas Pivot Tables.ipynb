# Basic Pivot Table Syntax

```python
pd.pivot_table(
    df,
    values="Revenue_INR",
    index="City",
    aggfunc="sum"
)
```

- values— Which column should be calculated?
- index— Which column form the rows?
- columns— Which column form separate columns?
- aggfunc— What calculation should be performed?
- fill_value— What value should replace combination mistakes?

```python
# Total Revenue by City
total_revenue=pd.pivot_table(df,
                            values="Revenue_INR",
                            index="City",
                            aggfunc="sum")
total_revenue
```

```python
# Average revenue by city
avg_revenue=pd.pivot_table(df,
                          values="Revenue_INR",
                          index="City",
                          aggfunc="mean")
avg_revenue
```

```python
# Total quantity by product
total=pd.pivot_table(df,
                    values="Quantity",
                    index="Product",
                    aggfunc="sum")
total_revenue
```

```python
# Average Price by Product
avg_price=pd.pivot_table(df,
                        values="Unit_Price_INR",
                        index="Product",
                        aggfunc="sum")
avg_price
```

```python
# Order Count by City
count_order=pd.pivot_table(df,
                          values="Order_ID",
                          index="City",
                          aggfunc="count")
count_order
```

```python
# Revenue by City and Product
pd.pivot_table(df,
              values="Revenue_INR",
              index="City",
              columns="Product",
              aggfunc="sum",
              fill_value=0)
```

```python
# Quantity by City and Product
pd.pivot_table(df,
              values="Quantity",
              index="City",
              columns="Product",
              aggfunc="sum",
              fill_value=0)
```

```python
# Order Count by City and Status
pd.pivot_table(df,
              values="Order_ID",
              index="City",
              columns="Order_Status",
              aggfunc="count",
              fill_value=0)
```

```python
# Revenue by category and city
pd.pivot_table(df,
              values="Revenue_INR",
              index="City",
              columns="Category",
              aggfunc="sum",
              fill_value=0)
```

```python
# Quantity by Product and Status
pd.pivot_table(df,
              values="Quantity",
              index="Product",
              columns="Order_Status",
              aggfunc="sum",
              fill_value=0)
```

```python
# Total and average revenue by city
pd.pivot_table(df,
              values="Revenue_INR",
              index="City",
              aggfunc=["mean","sum"])
```

```python
# Multiple Calculations by Product
pd.pivot_table(
    df,
    values=["Quantity", "Revenue_INR"],
    index="Product",
    aggfunc={
        "Quantity": "sum",
        "Revenue_INR": ["sum", "mean"]})
```

```python
# Revenue and quantity by city and product
pd.pivot_table(
    df,
    values=["Revenue_INR", "Quantity"],
    index="City",
    columns="Product",
    aggfunc="sum",
    fill_value=0)
```

```python
# Average quantity by city and product
pd.pivot_table(
    df,
    values="Quantity",
    index="City",
    columns="Product",
    aggfunc="mean",
    fill_value=0)
```

```python
# Count unique products by city
pd.pivot_table(
    df,
    values="Product",
    index="City",
    aggfunc="nunique")
```

```python
# Maximum revenue by city
pd.pivot_table(
    df,
    values="Revenue_INR",
    index="City",
    aggfunc="max")
```

```python
# Minimum Product Price by Category
pd.pivot_table(
    df,
    values="Unit_Price_INR",
    index="Category",
    aggfunc="min")
```

```python
# Revenue by city, product and order status
pd.pivot_table(
    df,
    values="Revenue_INR",
    index=["City", "Product"],
    columns="Order_Status",
    aggfunc="sum",
    fill_value=0)
```

```python
# 19 — Add row and column fulls
pd.pivot_table(
    df,
    values="Revenue_INR",
    index="City",
    columns="Product",
    aggfunc="sum",
    fill_value=0,
    margins=True,
    margins_name="Grand Total")
```

```python
# 20 — Create a full sales summary
sales_pivot = pd.pivot_table(
    df,
    values=["Revenue_INR", "Quantity", "Order_ID"],
    index="City",
    columns="Category",
    aggfunc={
        "Revenue_INR": "sum",
        "Quantity": "sum",
        "Order_ID": "count"
    },
    fill_value=0)

sales_pivot
```
