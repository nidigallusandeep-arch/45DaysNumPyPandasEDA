```python
numeric_df.corr(method="pearson")
```

```python
numeric_df.corr(method="kendall")
```

```python
corr_matrix = numeric_df.corr()
corr_pairs = corr_matrix.unstack()
corr_pairs = corr_pairs[
    corr_pairs.index.get_level_values(0) !=
    corr_pairs.index.get_level_values(1)]
corr_pairs.sort_values(ascending=False).head(10)
```

```python
corr_pairs.sort_values().head(10)
```

```python
corr_matrix["Revenue_INR"].sort_values(ascending=False)
```

```python
corr_matrix["Profit_INR"].sort_values(ascending=False)
```

```python
high_corr = corr_matrix[(corr_matrix > 0.7) | (corr_matrix < -0.7)]
high_corr
```

```python
target_corr = numeric_df.corr()["Revenue_INR"].drop("Revenue_INR")
print(target_corr.sort_values(ascending=False))
```
