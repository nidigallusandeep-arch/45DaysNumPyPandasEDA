```python
import numpy as np
students = np.array([
    "Sandeep", "Ravi", "Asha",
    "Neha", "Kiran", "Priya"])
marks = np.array([
    [85, 90, 95],
    [70, 75, 80],
    [92, 88, 96],
    [60, 65, 70],
    [np.nan, 80, 75],
    [78, 85, 82]])

subjects = np.array(["Python", "SQL", "NumPy"])
print("Students:", students)
print("Marks:\n", marks)
print("Shape:", marks.shape)
```

```python
print("Number of students:", len(students))
print("Number of subjects:", len(subjects))
print("Total marks entries:", marks.size)
print("Number of dimensions:", marks.ndim)
print("Data type:", marks.dtype)
```

```python
# 3. Handle missing values
missing_count=np.sum(np.isnan(marks))
print(missing_count)
```

```python
subject_averages = np.nanmean(marks, axis=0)
for i in range(len(subjects)):
    print(subjects[i], "Average:", subject_averages[i])
```

```python
# 4. Calculate each student's average
student_averages = np.nanmean(marks, axis=1)
for i in range(len(students)):
    print(students[i], "Average:", round(student_averages[i], 2))
```

```python
# 5. Find the topper and rank students
ranking = np.argsort(student_averages)[::-1]
print("Student Rankings:")
for rank, index in enumerate(ranking, start=1):
    print(
        rank,
        students[index],
        round(student_averages[index], 2))
```

```python
topper_index = np.argmax(student_averages)
print("Topper:", students[topper_index])
print("Average:", student_averages[topper_index])
```

```python
# 6. Analyze pass and fail counts
valid_marks = ~np.isnan(marks)
pass_mask = valid_marks & (marks >= 40)
fail_mask = valid_marks & (marks < 40)
print("Passed subject exams:", np.sum(pass_mask))
print("Failed subject exams:", np.sum(fail_mask))
print("Missing marks:", np.sum(~valid_marks))
```

```python
high_performers = students[student_averages >= 80]
print("High-performing students:", high_performers)
```

```python
# 7. Find the highest mark in each subject
highest_marks = np.nanmax(marks, axis=0)
for i in range(len(subjects)):
    print(subjects[i], "Highest mark:", highest_marks[i])
```

```python
for col in range(marks.shape[1]):
    topper_index = np.nanargmax(marks[:, col])
    print(
        subjects[col],
        "Topper:", students[topper_index],
        "Marks:", marks[topper_index, col])
```

```python
# 8. Filter and slice the data
print(marks[:3, :])
```

```python
# Select SQL marks for every student:
sql_marks = marks[:, 1]
print(sql_marks)
```

```python
#Find all recorded marks greater than 80:
high_marks = marks[marks > 80]
print(high_marks)
```

```python
# Find students whose average is between 75 and 90, inclusive:
selected = students[
    (student_averages >= 75) &
    (student_averages <= 90)]
print(selected)
```

```python
# 9. Count unique marks and their frequencies
recorded_marks = marks[~np.isnan(marks)]
values, counts = np.unique(
    recorded_marks,
    return_counts=True)
for value, count in zip(values, counts):
    print(value, "appears", count, "time(s)")
```
