```python
sns.histplot(data=df, x="Revenue_INR", bins=15)
plt.title("Revenue Distribution")
plt.show()
```

```python
sns.histplot(
    data=df,
    x="Revenue_INR",
    kde=True,
    bins=15)
plt.show()
```

```python
sns.kdeplot(data=df, x="Customer_Age", fill=True)
plt.title("Customer Age Distribution")
plt.show()
```

```python
sns.boxplot(data=df, x="Revenue_INR")
plt.show()
```

```python
sns.violinplot(
    data=df,
    x="Category",
    y="Revenue_INR")
plt.xticks(rotation=30)
plt.show()
```

```python
sns.ecdfplot(data=df, x="Revenue_INR")
plt.title("Revenue Cumulative Distribution")
plt.show()
```
