```python
inner_df = pd.merge(
    orders, customers,
    on="Customer_ID",
    how="inner")
inner_df.head()
```

```python
left_df = pd.merge(
    orders, customers,
    on="Customer_ID",
    how="left"
)
left_df.head()
```

```python
right_df = pd.merge(
    orders, customers,
    on="Customer_ID",
    how="right"
)
right_df.head()
```

```python
outer_df = pd.merge(
    orders, customers,
    on="Customer_ID",
    how="outer"
)
outer_df.head()
```

```python
result = pd.merge(
    orders,
    customers[["Customer_ID", "Customer_Name", "City"]],
    on="Customer_ID",
    how="left"
)
result.head()
```

```python
orders_copy = orders.rename(
    columns={"Customer_ID": "Order_Customer_ID"})

result = pd.merge(
    orders_copy,
    customers,
    left_on="Order_Customer_ID",
    right_on="Customer_ID",
    how="inner")
result.head()
```

```python
product_result = pd.merge(
    orders,
    products,
    on="Product",
    how="left")
product_result.head()
```

```python
result = pd.merge(
    orders, customers,
    on="Customer_ID",
    how="left"
)

result = pd.merge(
    result, products,
    on="Product",
    how="left"
)

result.head()
```

```python
result = pd.merge(
    orders, customers,
    on="Customer_ID",
    how="left",
    indicator=True
)

result[result["_merge"] == "left_only"]
```

```python
result = pd.merge(
    customers, orders,
    on="Customer_ID",
    how="left",
    indicator=True)

result[result["_merge"] == "left_only"]
```
