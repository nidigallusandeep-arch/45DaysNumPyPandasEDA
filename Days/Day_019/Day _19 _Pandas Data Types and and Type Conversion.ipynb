```python
df["City"]=df["City"].fillna(df["City"].mode()[0])
df["Monthly_Salary_INR"]=df["Monthly_Salary_INR"].fillna(df["Monthly_Salary_INR"].mean())
df["Performance_Rating"]=df["Performance_Rating"].fillna(df["Performance_Rating"].mean())
df["Joining_Date"]=df["Joining_Date"].fillna(df["Joining_Date"].mode()[0])
```

```python
# Convert a The column Using astype()
df["Age"]=df["Age"].astype("int64")
print(df["Age"].astype)
```

```python
# Convert a text column to to Pandas string type:
df["City"]=df["City"].astype("string")
print(df["City"].astype)
```

```python
# Convert joining Date Using to_datetime()
df["Joining_Date"]=pd.to_datetime(df["Joining_Date"],errors="coerce")
print(df["Joining_Date"].astype)
```

```python
df["Joining_Year"] = df["Joining_Date"].dt.year
print(df[["Employee_Name", "Joining_Date", "Joining_Year"]].head())
```

```python
# Convert salary to numeric
df["Monthly_Salary_INR"] = pd.to_numeric(df["Monthly_Salary_INR"],errors="coerce")
print(df["Monthly_Salary_INR"].dtype)
```

```python
print(df["Monthly_Salary_INR"].mean())
```

```python
# Salary analysis by department
avg_salary=df.groupby("Department")["Monthly_Salary_INR"].mean()
avg_salary
```

```python
# Find employees earning more than ₹40,000:
salary=df[df["Monthly_Salary_INR"]>40000].head()
print(salary[["Employee_Name","Department","Monthly_Salary_INR"]])
```
