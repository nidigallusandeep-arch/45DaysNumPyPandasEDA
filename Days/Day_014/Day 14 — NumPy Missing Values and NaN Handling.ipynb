# NumPy Missing Values & NaN]
- NaN means Not a Number. In NumPy, it is commonly used to represent a missing or undefined numerical value.

```python
import numpy as np
arr=np.array([10,20,np.nan,30,40,np.nan,50])
print(arr)
```

# 2. Find missing values using np.isnan()
- True means the value is NaN; False means it is not NaN.

```python
import numpy as np
arr=np.array([10,20,30,np.nan,40,50,np.nan,60])
result=np.isnan(arr)
print(result)          
```

```python
missing_count = np.sum(np.isnan(arr))
print("Missing values:", missing_count)
```

```python
valid_count = np.sum(~np.isnan(arr))
print("Valid values:", valid_count)
```

# 3. What happens when we calculate the mean
- Normally, np.mean() returns NaN if the array contains NaN.
- To calculate the mean while ignoring missing values, use np.nanmean()

```python
import numpy as np
arr=np.array([10,20,30,np.nan,40,50,np.nan,60])
print(np.mean(arr))
```

```python
print(np.nanmean(arr))
```

```python
arr = np.array([10, 20, np.nan, 40, 50])
print("Mean:", np.nanmean(arr))
print("Median:", np.nanmedian(arr))
print("Sum:", np.nansum(arr))
print("Minimum:", np.nanmin(arr))
print("Maximum:", np.nanmax(arr))
print("Standard deviation:", np.nanstd(arr))
print("Variance:", np.nanvar(arr))
```

# 5. Replace NaN with a specific value

```python
arr=np.array([10,20,30,np.nan,40,50,np.nan,60,70])
arr[np.isnan(arr)]=0
print(arr)
```

```python
arr=np.array([10,20,np.nan,30,40,np.nan,50,60])
arr[np.isnan(arr)]=1
print(arr)
```

# 6. Replace NaN with the mean

```python
arr=np.array([10,20,np.nan,30,40,np.nan,60,70])
mean_value=np.nanmean(arr)
arr[np.isnan(arr)]=mean_value
print(arr)
```

```python
arr=np.array([10,20,30,np.nan,40,50,np.nan])
mean_value=np.nanmean(arr)
arr[np.isnan(arr)]=mean_value
print(arr)
```

# 7. Replace NaN with the median

```python
arr=np.array([10,20,30,np.nan,40,50,np.nan,60])
median_value=np.nanmean(arr)
arr[np.isnan(arr)]=median_value
print(arr)
```

```python
arr=np.array([20,30,50,np.nan,40,60])
median_value=np.nanmean(arr)
arr[np.isnan(arr)]=median_value
print(arr)
```

```python
salaries = np.array([25000, 30000, np.nan, 40000, np.nan, 35000])
print("Missing salaries:", np.sum(np.isnan(salaries)))
print("Available salaries:", np.sum(~np.isnan(salaries)))
print("Average salary:", np.nanmean(salaries))
```

```python
average_salary = np.nanmean(salaries)
salaries[np.isnan(salaries)] = average_salary
print(salaries)
```

```python
marks = np.array([
    [80, 90, np.nan],
    [70, np.nan, 85],
    [60, 75, 95]])

print("Missing values:")
print(np.isnan(marks))
print("Mean of each column:")
print(np.nanmean(marks, axis=0))
print("Mean of each row:")
print(np.nanmean(marks, axis=1))
```
