# flatten()
- flatten() converts a multi-dimensional array into 1D.

```python
arr2 = np.array([
    [10, 20, 30],
    [40, 50, 60]])

flat = arr2.flatten()
print(flat)
```

```python
print(flat.shape)
```

# ravel()

# resize()
- resize() can change the size of an array

```python
arr=np.array([10,20,30,40,50,60])
arr.resize(2,2)
print(arr)
```

```python
arr=np.array([10,20,30,40,50])
arr.resize(3,2)
print(arr)
```
