# What are Apply, Map and Lambda?
- apply(): --Applies a function to a Series or DataFrame
- map()	: -- Maps values in a Series using a function or dictionary
- lambda :--Creates a small anonymous function

```python
square = lambda x: x ** 2

square(5)
```

```python
# Add two numbers
add = lambda a, b: a + b#
add(10, 20)
```

```python
# Example 3 — Check even or odd
check = lambda x: "Even" if x % 2 == 0 else "Odd"
check(10)
```

```python
# Example 4 — Find the largest number
maximum = lambda a, b: a if a > b else b
maximum(20, 30)
```

```python
# Example 5 — Convert employee names to uppercase
df["Employee_Name"] = df["Employee_Name"].apply(
    lambda x: x.strip().title())

df.head()
```

```python
# Example 6 — Annual salary calculation
df["Annual_Salary"] = df["Monthly_Salary_INR"].apply(
    lambda x: x * 12)
df.head()
```

```python
# Example 8 — Experience classification
df["Experience_Level"] = df["Experience_Years"].apply(
    lambda x: "Fresher" if x <= 1
    else "Mid-Level" if x <= 5
    else "Experienced")
df.head()
```

```python
# Example 9 — Performance classification
df["Performance_Status"] = df["Performance_Score"].apply(
    lambda x: "Excellent" if x >= 4.5
    else "Good" if x >= 3.5
    else "Average")
df.head()
```

```python
# Example 10 — Attendance classification
df["Attendance_Status"] = df["Attendance_Percent"].apply(
    lambda x: "Good" if x >= 90 else "Low")
```

```python
df.head()
```

```python
# Example 11 — Department name mapping
df["Department"] = df["Department"].str.strip().str.title()

df["Department_Name"] = df["Department"].map({
    "Sales": "Sales Department",
    "It": "Information Technology",
    "Hr": "Human Resources",
    "Finance": "Finance Department"
})
df.head()
```

```python
# Example 12 — City mapping
df["City_Code"] = df["City"].map({
    "Hyderabad": "HYD",
    "Bengaluru": "BLR",
    "Chennai": "CHE",
    "Vijayawada": "VJA"})
df.head()
```

```python
# Example 13 — Performance score mapping
df["Performance_Label"] = df["Performance_Status"].map({
    "Excellent": "A",
    "Good": "B",
    "Average": "C"
})
df.head()
```

```python
# Example 14 — Map experience level
df["Experience_Code"] = df["Experience_Level"].map({
    "Fresher": 1,
    "Mid-Level": 2,
    "Experienced": 3})
df.head()
```

```python
# Example 15 — Create salary category
df["Salary_Category"] = df["Monthly_Salary_INR"].apply(
    lambda x: "High" if x >= 60000
    else "Medium" if x >= 35000
    else "Low")
```
