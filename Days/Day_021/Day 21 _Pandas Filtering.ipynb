# Filter using multiple conditions
- Use & for AND and | for OR. Put each condition inside parentheses.

```python
# Hyderabad laptop orders
result=df[(df["City"]=="Hyderabad")&(df["Product"]=="Laptop")]
result
```

```python
# Laptops or monitors
result=df[(df["Product"]=="Monitor")|(df["Product"]=="Laptop")].head()
result
```

```python
# Delivered orders from Bengaluru
resul=df[(df["City"]=="Bengaluru")&(df["Order_Status"]=="Delivered")]
result
```

```python
# Expensive electronics
result=df[(df["Category"]=="Electronics")&(df["Unit_Price_INR"]>20000)].head()
result
```

# Use isin() to match multiple values
- Suppose your manager wants orders from Hyderabad, Bengaluru, and Chennai.

```python
cites=["Hyderabad","Bengalure","Chennai","Pune"]
result=df[df["City"].isin(cites)].head()
result
```

```python
products = ["Laptop", "Monitor"]
result = df[df["Product"].isin(products)].head()
result
```

```python
result = df[~df["Product"].isin(products)].head()
result
```

# Use between() for ranges
- Find products priced between ₹1,000 and ₹20,000, including both boundaries.

```python
result=df[df["Unit_Price_INR"].between(1000,20000)].head()
print(result[["Product", "Unit_Price_INR"]])
```

```python
# Find orders with quantity between 2 and 3:
result=df[df["Quantity"].between(2,3)].head()
result
```

# Handle missing values during filtering

```python
# Find rows with missing customer names:
df[df["Customer_Name"].isna()]
```

```python
# Find rows with available customer names:
df[df["Customer_Name"].notna()].head()
```

```python
# Find rows with missing quantities:
df[df["Quantity"].isna()]
```
