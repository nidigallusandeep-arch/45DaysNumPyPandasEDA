# Day 02 — NumPy Array Creation

```python
import numpy as np
```

# np.zeros()
- Creates an array filled with 0.

```python
arr=np.zeros(5)
print(arr)
```

```python
arr=np.zeros(7)
arr
```

```python
arr=np.zeros(4)
print(arr)
```

# 2D zeros

```python
arr=np.zeros((3,6))
print(arr)
```

```python
arr=np.zeros((4,7))
arr
```

# 3. np.ones()
- Creates an array filled with 1.

```python
arr=np.ones(3)
arr
```

```python
arr=np.ones(4)
arr
```

# 2D:

```python
arr=np.ones((4,5))
arr
```

```python
arr=np.ones((3,7))
arr
```

# 3. np.arange()
- Creates numbers within a given range.

```python
arr=np.arange(1,11)
arr
```

- With step:

```python
arr=np.arange(2,21,2)
arr
```

# 4. np.linspace()
- Creates evenly spaced numbers.

```python
arr=np.linspace(0,10,5)
arr
```

```python
arr=np.linspace(1,5,3)
arr
```

- Difference:
- arange  → step-based
- linspace → number-of-values based

# 5.np.full()
- Creates an array with the same value.

```python
arr=np.full(5,28)
arr
```

```python
arr=np.full(5,7)
arr
```

```python
arr=np.full(7,26)
arr
```

# 2D

```python
arr=np.full((3,3),7)
arr
```

```python
arr=np.full((4,4),28)
arr
```

```python
arr=np.full((3,3),10)
arr
```

# 6. np.eye()
- Creates an identity matrix.

```python
arr=np.eye(4)
arr
```

```python
arr=np.eye(3)
arr
```
