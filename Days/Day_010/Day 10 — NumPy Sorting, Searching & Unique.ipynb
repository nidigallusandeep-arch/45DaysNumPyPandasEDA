# np.argsort()
- argsort() gives the index positions that would sort the array.

```python
arr = np.array([50, 20, 80, 10, 40])
result=np.argsort(arr)
print(result)
```

# Use argsort() to sort another array

```python
name=np.array(["A","B","C","D"])
marks=np.array([80,60,76,98])
result=np.argsort(marks)
print(result)

print(name[result])
print(marks[result])
```

# Descending with argsort()

```python
indices = np.argsort(marks)[::-1]

print(name[indices])
print(marks[indices])
```

# np.unique() 
- unique() gives only unique values.

```python
arr=np.array([10,20,10,20,30,304,10,30,40,50,40,60])
result=np.unique(arr)
print(result)
```

# Count unique values

```python
arr=np.array(["Python", "SQL", "Python", "Excel", "SQL", "Python"])
values,counts=np.unique(arr,return_counts=1)
print("Values:",values)
print("counts:",counts)
```

# Find positions using np.where()

```python
arr=np.array([10,20,30,10,40,50])
print(np.where(arr>25))
```

```python
result=np.where(arr>20)[0]
result
```

# np.searchsorted()
- Used to find where a value should be inserted in a sorted array.

```python
arr=np.array([1,20,30,40,20,40,50,50,60])
result=np.searchsorted(arr,25)
print(result)
```

# Search multiple values

```python
arr=np.array([10,20,30,40,50,60,90])
result=np.searchsorted(arr,[10,60,90])
print(result)
```

# Sorting a 2D array

```python
arr=np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]])
print(arr)
```

```python
# Sort columns using axis=0
print(np.sort(arr, axis=0))
```

```python
# Sort rows using axis=1
print(np.sort(arr, axis=1))
```

```python
# Real Example — Student Ranking
students = np.array([
    "Sandeep",
    "Ravi",
    "Asha",
    "Neha",
    "Kiran"])
marks = np.array([78, 92, 85, 95, 70])
print(students)
print(marks)
```

```python
ranking = np.argsort(marks)[::-1]
print(students[ranking])
print(marks[ranking])
```

# Real Example — Product Prices

```python
products = np.array([
    "Laptop",
    "Mouse",
    "Keyboard",
    "Monitor",
    "Headphones"])

prices = np.array([
    60000,
    1200,
    2500,
    15000,
    3000])
```

```python
indices = np.argsort(prices)
print(products[indices])
print(prices[indices])
```
