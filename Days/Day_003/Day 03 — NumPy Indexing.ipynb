# 5. 2D Array Indexing

```python
import numpy as np
arr2=np.array([
    [10,20,30,],
    [40,50,60],
    [70,80,90]])
print(arr2)
```

```python
# Access 10
print(arr2[0,0])
```

```python
# Access 50
print(arr2[1,1])
```

```python
# Access 90
print(arr2[2,2])
```

# 6. Access a complete row
- First row:

```python
print(arr2[0])
```

```python
# Second row:
print(arr2[1])
```

```python
print(arr2[-1])
```

# 7. Access a complete column
- First column:

```python
print(arr2[:,0])
```

```python
# Second column:
print(arr2[:,1])
```

```python
print(arr2[:,-1])
```
