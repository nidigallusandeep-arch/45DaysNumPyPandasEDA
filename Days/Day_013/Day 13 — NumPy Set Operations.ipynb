```python
import numpy as np
arr1=np.array([10,20,30,40,50,60])
arr2=np.array([20,10,70,30,80,90])
print(arr1)
print(arr2)
```

# 2. np.union1d() — Combine unique values
- It returns all unique values from both arrays.

```python
import numpy as np
arr1=np.array([10,20,30,40,50,60])
arr2=np.array([20,10,70,30,80,90])
result=np.union1d(arr1,arr2)
print(result)
```

# 3. np.intersect1d() — Find common values
- Returns the values present in both arrays.

```python
import numpy as np
arr1=np.array([1,2,3,4,5,6])
arr2=np.array([2,4,5,7,8,9])
result=np.intersect1d(arr1,arr2)
print(result)
```

```python
session1 = np.array([101, 102, 103, 104, 105])
session2 = np.array([103, 104, 105, 106, 107])
common_students = np.intersect1d(session1, session2)
print(common_students)
```

# 4. np.setdiff1d() — Find values in the first array only
- Returns values present in arr1 but not in arr2

```python
import numpy as np
arr1=np.array([10,20,30,40,50,60])
arr2=np.array([20,10,70,30,80,90])
result=np.setdiff1d(arr1,arr2)
print(result)
```

# np.setxor1d() — Find values that are not common
- Returns values that appear in either array, but not in both.

```python
import numpy as np
arr1=np.array([10,20,30,40,50,60])
arr2=np.array([20,10,70,30,80,90])
result=np.setxor1d(arr1,arr2)
print(result)
```

```python
import numpy as np
warehouse_a = np.array([1001, 1002, 1003, 1004])
warehouse_b = np.array([1003, 1004, 1005, 1006])
print("All products:",
      np.union1d(warehouse_a, warehouse_b))
print("Available in both:",
      np.intersect1d(warehouse_a, warehouse_b))
print("Only in warehouse A:",
      np.setdiff1d(warehouse_a, warehouse_b))
print("Available in only one warehouse:",
      np.setxor1d(warehouse_a, warehouse_b))
```
