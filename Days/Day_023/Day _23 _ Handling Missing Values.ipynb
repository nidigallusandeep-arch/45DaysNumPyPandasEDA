```python
# Remove rows containing any missing value
df_clean = df.dropna()
print(df_clean.shape)
```

```python
# Remove rows only when Customer_Name is missing
df_clean = df.dropna(subset=["Customer_Name"])
```

```python
# Remove columns that contain missing values
df_clean = df.dropna(axis=1)
```

```python
# fillna() — Fill missing values
df["Customer_Name"] = df["Customer_Name"].fillna("Unknown")
```

```python
# Fill missing cities with the most frequent city
most_common_city = df["City"].mode()[0]
df["City"] = df["City"].fillna(most_common_city)
```

```python
#Fill missing quantity with median
median_quantity = df["Quantity"].median()

df["Quantity"] = df["Quantity"].fillna(
    median_quantity
)
```

```python
# Fill missing price with median
median_price = df["Unit_Price_INR"].median()
df["Unit_Price_INR"] = df["Unit_Price_INR"].fillna(
    median_price)
```

```python
df[df["Payment_Method"].isna()]
```

```python
df["Payment_Method"] = df["Payment_Method"].fillna(
    df["Payment_Method"].mode()[0]
)
```

```python
print(df["Payment_Method"].isna().sum())
```

```python
df["Revenue_INR"] = df["Quantity"] * df["Unit_Price_INR"]
```
