# What is Crosstab?
- pd.crosstab() creates a frequency table that shows how often categories occur together.

# For example, a company manager wants to know:
- How many orders came from each city?
- How many orders are Delivered, Cancelled, or Returned?
- Which city has the most Delivered orders?

# Basic syntax:
pd.crosstab(
    index=df["City"],
    columns=df["Order_Status"]
)

- index — categories shown as rows.
- columns — categories shown as columns.
- By default, the table counts how many times each combination occurs.

```python
# Category count in each city
pd.crosstab(["City"],["Category"])
```

```python
# Add row and column totals
pd.crosstab(df["City"],["Order_Status"],margins=True)
```

- margins=True adds totals for each row and column.

```python
# Add a custom total name
pd.crosstab(df["City"],df["Order_Status"],margins=True,margins_name="Grand Total")
```

# Percentages and normalization

```python
# Row-wise percentages
pd.crosstab(df["City"],df["Order_Status"],normalize="index").round(2) * 100
```

- Shows the percentage distribution of order statuses within each city.

```python
# Column-wise percentages
pd.crosstab(df["City"],df["Order_Status"],normalize="columns").round(2) * 100
```

- Shows how each order status is distributed across cities.


```python
# Overall percentages
pd.crosstab(df["City"],df["Order_Status"],normalize="all").round(3) * 100
```

- Shows each city-status combination as a percentage of all rows.

```python
# Count unique order IDs
pd.crosstab(
    df["City"],
    df["Order_Status"],
    values=df["Order_ID"],
    aggfunc="nunique").fillna(0).astype(int)
```

```python
# Total quantity by city and product
pd.crosstab(
    df["City"],
    df["Product"],
    values=df["Quantity"],
    aggfunc="sum"
).fillna(0)
```

```python
# Total revenue by city and product
pd.crosstab(
    df["City"],
    df["Product"],
    values=df["Revenue_INR"],
    aggfunc="sum"
).fillna(0)
```

```python
# Average revenue by city and product
pd.crosstab(
    df["City"],
    df["Product"],
    values=df["Revenue_INR"],
    aggfunc="mean"
).fillna(0)
```

```python
# Maximum quantity by city and product
pd.crosstab(
    df["City"],
    df["Product"],
    values=df["Quantity"],
    aggfunc="max"
).fillna(0)
```

```python
# City and category with totals
pd.crosstab(
    df["City"],
    df["Category"],
    margins=True,
     margins_name="Total"
)
```

```python
# Product and order status with totals
pd.crosstab(df["Product"],df["Order_Status"],margins=True)
```

```python
# Count each category by city, normalized
pd.crosstab(
    df["City"],
    df["Category"],
    normalize="index").round(2) * 100
```

```python
# 9 — Filter Delivered orders and count products by city
delivered = df[df["Order_Status"] == "Delivered"]
pd.crosstab(
    delivered["City"],
    delivered["Product"],
    margins=True)
```

```python
# Create a city-wise order status report
status_report = pd.crosstab(
    index=df["City"],
    columns=df["Order_Status"],
    margins=True,
    margins_name="Total")

status_report
```
