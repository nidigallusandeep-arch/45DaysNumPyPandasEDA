# sort_index() — Sort by index
- sort_values() sorts using column values. sort_index() sorts using row or column index labels.

```python
# Sort rows by index
df.sort_index(ascending=True).head()
```

```python
# Reverse row-index order
df.sort_index(ascending=False).head()
```

```python
df.sort_index(axis=1)
```

# nlargest() and nsmallest()
- These are convenient shortcuts when you need only the top or bottom values.

```python
# Top 5 highest-revenue orders
df.nlargest(5, "Revenue_INR")
```

```python
# 5 lowest-priced orders
df.nsmallest(5, "Unit_Price_INR")
```

# Handle ties and missing values

```python
# Highest price first; break ties by product name
df.sort_values(
    by=["Unit_Price_INR", "Product"],
    ascending=[False, True])
```

```python
# Put missing prices at the end
df.sort_values(
    by="Unit_Price_INR",
    na_position="last")
```
