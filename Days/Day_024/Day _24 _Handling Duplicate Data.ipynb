# duplicated() — Find duplicate rows
- False means the row is not a duplicate of an earlier row.
- True means the row repeats an earlier row across all columns.

```python
df.duplicated()
```

```python
# Count duplicate rows
print(df.duplicated().sum())
```

```python
# Display duplicate rows
df[df.duplicated()]
```

- This shows the repeated rows, excluding the first occurrence.

# drop_duplicates() — Remove duplicates

```python
df_clean=df.drop_duplicates()
print("Before:", df.shape)
print("After:", df_clean.shape)
```

```python
# Remove duplicates using a specific column
df_clean = df.drop_duplicates(subset=["Order_ID"])
```

```python
# Keep the last occurrence
df_clean = df.drop_duplicates(keep="last")
```

# Options:
- keep="first" — keep the first occurrence (default).
- keep="last" — keep the last occurrence.
- keep=False — remove every row that belongs to a duplicate group.

```python
# Check duplicates using multiple columns
df.duplicated(subset=["Customer_Name", "Product"])
```

```python
# Find duplicate order IDs
df[df.duplicated(subset=["Order_ID"], keep=False)]
```

```python
# Remove exact duplicates and reset the index
df_clean = (df.drop_duplicates().reset_index(drop=True))
print(df_clean.shape)
```

```python
# Check whether duplicates remain
print(df_clean.duplicated().sum())
```
