#  NumPy Random Functions

```python
# 1. Import NumPy
import numpy as np
```

```python
# 2. np.random.randint() — Random integers
import numpy as np
arr = np.random.randint(1, 100, size=5)
print(arr)
```

```python
# 3. Generate a 1D array
arr=np.random.randint(3,10,size=10)
print(arr)
```

# 4. Generate a 2D array

```python
arr=np.random.randint(1,100,size=(3,4))
print(arr)
```

# 5. np.random.rand() — Random decimal numbers
- Generates random decimal values between 0 and 1.

```python
arr=np.random.rand(5)
print(arr)
```

```python
# Generate a 2D array:
arr=np.random.rand(3,4)
print(arr)
```

# 6. np.random.randn() — Standard normal distribution
- Generates values from a standard normal distribution, with mean approximately 0 and standard deviation approximately 1.

```python
arr=np.random.randn(5)
print(arr)
```

```python
# 7. np.random.random() — Random decimal values
arr=np.random.random(5)
print(arr)
```

# 8. np.random.choice() — Select random values

```python
marks=np.array([45,50,60,55,70,80,70,92])
result=np.random.choice(marks,size=5)
print(result)
```

```python
result = np.random.choice(marks, size=3, replace=False)
print(result)
```

# 9. np.random.shuffle() — Shuffle an array

```python
arr=np.array([10,20,30,40,50])
np.random.shuffle(arr)
print(arr)
```

# 10. np.random.permutation() — Random rearrangement

```python
arr=np.array([10,20,30,40,50,60])
result=np.random.permutation(arr)
print(result)
```

# 11. np.random.seed() — Reproducible results

```python
np.random.seed(42)
arr = np.random.randint(1, 100, size=5)
print(arr)
```

```python
rng = np.random.default_rng(42)
arr = rng.integers(1, 100, size=5)
print(arr)
```

# 12. Real-life example — Student marks analysis

```python
import numpy as np
rng = np.random.default_rng(42)
marks = rng.integers(0, 101, size=10)
print("Student Marks:", marks)
print("Average Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
print("Passed Students:", np.sum(marks >= 40))
print("Failed Students:", np.sum(marks < 40))
```

# 13. Real-life example — Sales data

```python
rng = np.random.default_rng(42)
sales = rng.integers(1000, 10001, size=7)
print("Weekly Sales:", sales)
print("Total Sales:", np.sum(sales))
print("Average Sales:", np.mean(sales))
print("Maximum Sales:", np.max(sales))
print("Minimum Sales:", np.min(sales))
```
