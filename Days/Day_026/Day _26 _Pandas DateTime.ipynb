# Convert text to datetime
- Use pd.to_datetime():

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
print(df.dtypes)
```

```python
df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Day"] = df["Order_Date"].dt.day
print(df[["Order_Date", "Year", "Month", "Day"]].head())
```

```python
df["Month_Name"] = df["Order_Date"].dt.month_name()
df["Day_Name"] = df["Order_Date"].dt.day_name()
df["Quarter"] = df["Order_Date"].dt.quarter
```

# Filter data using dates
- Example 1: Orders after June 1, 2026

```python
result = df[df["Order_Date"] >= "2026-06-01"].head()
result
```

```python
# Example 2: Orders from March 1 to March 31
result = df[
    (df["Order_Date"] >= "2026-03-01") &
    (df["Order_Date"] < "2026-04-01")]

result
```

```python
# Example 3: Orders from October
october_orders = df[df["Order_Date"].dt.month == 10]
october_orders
```

# Calculate date differences
- Suppose we want to calculate how many days have passed since each order date, using October 24, 2026, as the reference date.

```python
reference_date = pd.Timestamp("2026-10-24")
df["Days_Since_Order"] = (reference_date - df["Order_Date"]).dt.days
print(df[["Order_Date", "Days_Since_Order"]].head())
```

# Format dates as strings

```python
df["Formatted_Date"] = df["Order_Date"].dt.strftime("%d-%m-%Y")
print(df[["Order_Date", "Formatted_Date"]].head())
```

```python
# Handle invalid or missing dates
dates = pd.Series([
    "2026-01-15",
    "not-a-date",
    None])

converted_dates = pd.to_datetime(
    dates,
    errors="coerce")
print(converted_dates)
```
