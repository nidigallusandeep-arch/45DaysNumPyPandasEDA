```python
# Example 1 — Combine January and February
result = pd.concat([jan, feb])
result.head()
```

```python
# Example 2 — Combine with reset index
result = pd.concat(
    [jan, feb],
    ignore_index=True)
result.head()
```

```python
# Example 3 — Combine three DataFrames
result = pd.concat(
    [jan, feb, mar],
    ignore_index=True)
result.head()
```

```python
# Example 5 — Combine only selected columns
result = pd.concat([
    jan[["Order_ID", "City", "Revenue_INR"]],
    feb[["Order_ID", "City", "Revenue_INR"]]
], ignore_index=True)
result.head()
```

```python
# Example 6 — Combine rows
pd.concat([jan, feb], axis=0, ignore_index=True).head()
```

```python
# Example 7 — Combine columns
pd.concat([jan, feb], axis=1)
```

- Note: axis=1 aligns rows by index. It does not match orders by Order_ID.

```python
# Example 8 — Combine selected columns side by side
pd.concat([
    jan[["Order_ID", "City"]],
    jan[["Product", "Revenue_INR"]]], axis=1)
```

```python
# Example 9 — Keep original index
pd.concat([jan, feb], ignore_index=False).head()
```

```python
# Example 10 — Add source labels using keys
result = pd.concat(
    [jan, feb],
    keys=["January", "February"])

result
```

```python
# Example 11 — Drop old index
result = pd.concat(
    [jan, feb],
    ignore_index=True)
result.head()
```

```python
# Example 12 — Inner join on columns
result = pd.concat(
    [jan, mar],
    join="inner",
    ignore_index=True)
result.head()
```

```python
# Example 13 — Outer join on columns
result = pd.concat(
    [jan, mar],
    join="outer",
    ignore_index=True)
result.head()
```

- join="outer" keeps all columns. Missing values are filled with NaN.

```python
# Example 14 — Check missing values after concat
result = pd.concat(
    [jan, mar],
    ignore_index=True)
result.isnull().sum()
```

```python
# Example 15 — Remove duplicate rows after concat
result = pd.concat(
    [jan, feb],
    ignore_index=True)

result = result.drop_duplicates()
```

```python
# Example 16 — Sort combined orders by revenue
result = pd.concat(
    [jan, feb],
    ignore_index=True
)

result.sort_values(
    "Revenue_INR",
    ascending=False
)
```

```python
# Example 17 — Filter delivered orders after concat
result = pd.concat(
    [jan, feb],
    ignore_index=True)
result[result["Order_Status"] == "Delivered"]
```

```python
# Example 18 — Calculate total monthly revenue
result = pd.concat(
    [jan, feb],
    ignore_index=True)

result.groupby("City")["Revenue_INR"].sum()
```

```python
# Example 19 — Create month-wise DataFrame labels
jan_copy = jan.copy()
feb_copy = feb.copy()
jan_copy["Month"] = "January"
feb_copy["Month"] = "February"
result = pd.concat(
    [jan_copy, feb_copy],
    ignore_index=True)

result.head()
```

```python
# Example 20 — Complete combined sales report
jan_copy = jan.copy()
feb_copy = feb.copy()

jan_copy["Month"] = "January"
feb_copy["Month"] = "February"
final_df = pd.concat(
    [jan_copy, feb_copy],
    ignore_index=True)
sales_report = final_df.groupby("Month").agg(
    Total_Orders=("Order_ID", "count"),
    Total_Quantity=("Quantity", "sum"),
    Total_Revenue=("Revenue_INR", "sum"))

sales_report
```
