# What is Univariate Analysis?
- Uni = One, Variate = Variable
- Univariate Analysis means analyzing only one column (variable) at a time.

# For example:
- Customer age distribution.
- Product order count.
- Customer rating distribution.
- Total quantity distribution.
- Revenue distribution.
# Why do we use it?
1. Understand data distribution.
2. Identify most frequent values.
3. Find average, median and mode.
4. Identify outliers.
5. Understand categorical and numerical variables.

# Types of Univariate Analysis
# 1. Numerical Univariate Analysis

- Example: Age, Salary, Revenue, Quantity.
- Common charts: Histogram, Boxplot, KDE plot.


# 2. Categorical Univariate Analysis

- Example: City, Product, Order Status.
- Common charts: Countplot, Bar chart, Pie chart.

# Numerical visualizations
- Example 11 — Age histogram

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Customer_Age"], bins=8)
plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# Example 12 — Revenue histogram
plt.figure(figsize=(8, 5))
sns.histplot(df["Revenue_INR"], bins=10, kde=True)
plt.title("Revenue Distribution")
plt.show()
```

```python
# Example 13 — Quantity boxplot
plt.figure(figsize=(8, 4))
sns.boxplot(x=df["Quantity"])
plt.title("Quantity Outlier Analysis")
plt.show()
```

```python
# Example 14 — Customer rating distribution
sns.histplot(
    df["Customer_Rating"],
    bins=8,
    kde=True)
plt.title("Customer Rating Distribution")
plt.show()
```

```python
# Example 15 — Revenue KDE plot
sns.kdeplot(
    data=df,
    x="Revenue_INR",
    fill=True)
plt.title("Revenue Density Distribution")
plt.show()
```

# Categorical visualizations
- Example 16 — Product countplot

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Product")
plt.title("Product Order Count")
plt.xticks(rotation=30)
plt.show()
```

```python
# Example 17 — City-wise order count
sns.countplot(data=df, x="City")
plt.title("Orders by City")
plt.xticks(rotation=30)
plt.show()
```

```python
# Example 18 — Order status bar chart
df["Order_Status"].value_counts().plot(
    kind="bar",
    figsize=(7, 4))
plt.title("Order Status Distribution")
plt.xlabel("Status")
plt.ylabel("Count")
plt.show()
```

```python
# Example 19 — Product pie chart
df["Product"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7, 7))

plt.title("Product Distribution")
plt.ylabel("")
plt.show()
```

```python
# Example 20 — Customer age boxplot
plt.figure(figsize=(8, 4))
sns.boxplot(y=df["Customer_Age"])
plt.title("Customer Age Outlier Detection")
plt.show()
```
