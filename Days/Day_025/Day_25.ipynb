# .str.lower() — Convert to lowercase

```python
df["Customer_Name"]=df["Customer_Name"].str.lower()
print(df["Customer_Name"])
```

- Example: "SANDEEP" → "sandeep"

```python
df["Email"]=df["Email"].str.lower()
print(df["Email"])
```

```python
df["City"]=df["City"].str.lower()
print(df["City"])
```

# .str.upper() — Convert to uppercase

```python
df["City"]=df["City"].str.upper()
print(df["City"])
```

- Example: "Hyderabad" → "HYDERABAD"

```python
df["Customer_Name"]=df["Customer_Name"].str.upper()
print(df["Customer_Name"])
```

```python
df["Product"]=df["Product"].str.upper()
print(df["Product"])
```

# .str.title() — Capitalize each word
- Example: "sandeep sharma" → "Sandeep Sharma"

- Note: Names with special capitalization, such as McDonald, may need custom handling

```python
df["Customer_Name"]=df["Customer_Name"].str.title()
print(df["Customer_Name"])
```

# .str.strip() — Remove extra spaces

- Example:

- "  sandeep " → "sandeep

```python
df["Customer_Name"]=df["Customer_Name"].str.strip()
print(df["Customer_Name"])
```

```python
df["City"]=df["City"].str.strip()
print(df["City"])
```

# .str.contains() — Search for text
-  Find customers whose names contain "sha"


```python
result = df[df["Customer_Name"].str.contains("sha", case=False, na=False)]

print(result)
```

# .str.replace() — Replace text

```python
df["Email"] = df["Email"].str.replace("EXAMPLE.COM","example.com",case=False,regex=False)
```

```python
df.head()
```

- Example: "RAVI@EXAMPLE.COM" → "RAVI@example.com"

```python
df["Order_Status"] = df["Order_Status"].str.lower()
```

- This converts "DELIVERED", "Delivered", and "delivered" to the same lowercase format.

# .str.split() — Split text into parts

```python
df["Customer_Name"].str.split()
```

- Example: "Priya Sharma" → ["Priya", "Sharma"]

```python
df["First_Name"] = (
    df["Customer_Name"]
    .str.strip()
    .str.split()
    .str[0])
print(df[["Customer_Name", "First_Name"]].head())
```

# .str.startswith() and .str.endswith()

```python
# Emails that start with "s"
df[df["Email"].str.startswith("s", na=False)]
```

```python
# Emails that end with ".com"
df[df["Email"].str.endswith(".com", na=False)]
```
