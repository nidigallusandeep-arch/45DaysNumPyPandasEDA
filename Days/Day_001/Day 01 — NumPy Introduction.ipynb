# 1. Create a NumPy array

```python
import numpy as np
```

```python
arr=np.array([10,20,30,40,50,])
```

```python
print("Array:",arr)
```

```python
# 2. Shape
print("Shape:",arr.shape)
```

```python
# 3. Size
print("Size:",arr.size)
```

```python
# 4. Number of dimensions
print("Dimensions:",arr.ndim)
```

```python
# 5. Data type
print("Data_type:",arr.dtype)
```
