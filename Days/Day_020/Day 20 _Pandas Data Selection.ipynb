# Select rows using iloc[]
- iloc[] selects rows and columns using integer positions, starting from zero.

```python
# First row
df.iloc[0]
```

```python
# First three rows
df.iloc[0:3]
```

```python
df.iloc[0:1]
```

```python
# First three rows and first two columns
df.iloc[0:3,0:2]
```

# Select data using loc[]
- loc[] selects data using row labels and column names.

```python
# First row by its label
df.loc[0]
```

```python
# Select customer and product columns
df.loc[:, ["Customer_Name", "Product"]].head()
```

```python
# Select rows with labels 0 through 4, inclusive
df.loc[0:4, ["Order_ID", "Product"]]
```

```python
expensive = df.loc[df["Unit_Price_INR"] > 10000,["Order_ID", "Product", "Unit_Price_INR"]]
expensive
```

```python
laptop_orders = df.loc[df["Product"] == "Laptop"]
laptop_orders
```

```python
# Find delivered orders from Hyderabad
result = df.loc[(df["City"] == "Hyderabad") &(df["Order_Status"] == "Delivered"),["Order_ID", "Customer_Name", "Product", "Order_Status"]]
result
```

```python
# Select orders paid through UPI
upi_orders = df.loc[df["Payment_Method"] == "UPI"].head()
upi_orders
```
