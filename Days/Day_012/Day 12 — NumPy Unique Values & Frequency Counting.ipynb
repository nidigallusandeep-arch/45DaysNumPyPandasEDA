# What is np.unique
- np.unique() returns only the distinct values from an array and removes duplicates.

```python
import numpy as np
arr=np.array([10,20,30,40,50,60,70,80,70,80])
result=np.unique([arr])
print(result)
```

```python
# Example 1: Student marks

marks=np.array([80,60,70,70,80,80,50,70,60])
print(np.unique(marks))
```

```python
# 2 Employee departments
departments = np.array(["IT", "HR", "IT", "Sales", "HR", "IT"])
print(np.unique(departments))
```

```python
# Example 3: Product categories

products = np.array(["Mobile", "Laptop", "Mobile","Tablet", "Laptop", "Mobile"])
print(np.unique(products))
```

# Count frequency using return_counts=True

```python
arr=np.array([10,20,30,10,30,40,50,40,10])
values,count=np.unique(arr,return_counts=True)
print("Values:",values)
print("Count:",count)
```

# Real-time example — Count student grades

```python
grades = np.array([
    "A", "B", "A", "C", "B",
    "A", "A", "C", "B", "A"])
grade_names, grade_counts = np.unique(grades,return_counts=True)
print("Grades:", grade_names)
print("Frequency:", grade_counts)
```

# Find the most frequent value

```python
values, counts = np.unique(grades, return_counts=True)
index = np.argmax(counts)
print("Most frequent grade:", values[index])
print("Frequency:", counts[index])
```

# Frequency counting with np.bincount()
- np.bincount() counts occurrences of non-negative integers, including integers that do not appear in the array.

```python
numbers = np.array([1, 2, 2, 3, 3, 3, 5])
counts = np.bincount(numbers)
print(counts)
```
